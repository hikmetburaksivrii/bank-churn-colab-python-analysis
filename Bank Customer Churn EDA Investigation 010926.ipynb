D. DEMOGRAFİ: AGE!

```python
px.histogram(df_bank_churn, x="age", color="churn", nbins=40, barmode="overlay",
             opacity=0.7,title="Age Distribution — Exited vs Retained Customers",
             labels={"age": "Age", "churn": "Churn"}).show()
# 'Nbins' (Kutu Sayısı), Histogramdaki Sürekli Verilerin (Örn. Yaş) Kaç Eşit Aralığa (Kutu/Bin) Bölüneceğini Belirler!
# Görsel Olarak Dikey Çubuklara Benzeseler De, Teknik Olarak Her Bir Yapı Veri Grubunu Temsil Eden Birer "Kutu"dur (Bin)!
# 'Barmode' ("Overlay"), Farklı Kategorilere Ait (Örn. Churn = 0 Ve Churn = 1) Çubukları Üst Üste (Çakışık) Ekran Üzerinde Çizer.
# 'Opacity' (Saydamlık), Üst Üste Binmiş Çubukların Arkasında Kalan Alanların Net Şekilde Görünebilmesi İçin Şeffaflık Oranını Ayarlar.
```

```python
#Verilen Sütuna Göre Müşterileri Gruplandırır Ve Her Grubun Müşteri Sayısını Ve Churn Oranını Hesaplar!
def churn_by(columns):
    result = (df_bank_churn.groupby(columns, observed=True)
              .agg(customer_count=("customer_id", "count"),
              churn_rate=("churn", "mean")).reset_index())
    result["churn_rate"] = (result["churn_rate"] * 100).round(1)
    return result
# 'Observed=True', Tanımlı Olmasına Rağmen Veride Hiç Bulunmayan Boş Kategorileri Tablodan Tamamen Çıkarır.
# 'Observed=False' İse Veride Hiç Müşteri Olmasa Bile O Kategoriyi 0 Müşteri Ve NaN Oranıyla Tabloya Dahil Eder.
# Örnek: Veride 51+ Yaşında Müşteri Yoksa, observed=True Bu Grubu Gizler; observed=False İse Müşteri Sayısını 0 Gösterir.
# 'Agg()' Fonksiyonu, Gruplanmış Veriler Üzerinde Birden Fazla İstatistiksel Hesaplamayı Tek Adımda Yapmayı Sağlar.
# 'Yeni Sutun Adi=("Mevcut Sutun", "Islem")' Yapısıyla, Hesaplanan Metriklere Doğrudan Yeni İsimler Atamamıza Olanak Tanır.
# Kod Özelinde 'AGG()', Her Grup İçin Hem Müşteri Sayısını ('count') Hem De Ayrılma Oranını ('mean') Aynı Anda Hesaplar.
```

```python
# Fonksiyonun Amacı, Verilen Bir DataFrame İçindeki Churn Rate (Müşteri Kayıp Oranını)
# Kategorik Değişkene Göre Sütun Grafik (Bar Chart) Olarak Göstermektir!
def draw_bar_chart(data_frame, x_col, title):
    # Küçük & Büyük Harf Fark Etmeksizin Sütun Adını Bulmak.
    # Her Sütun Adını Tek Tek Col Değişkenine Koyar ve Sütun Adının Küçük Harflerle Yazımı Kontrol Edilir.
    # Kullanıcıdan Gelen Kolon Adının, Büyük Küçük Harfleri Hatalı Yazılsa Bile Amaç DF'Deki Doğru Kolonla Eşleştirmek.
    # Kontrol Sonucunda İstenilen Sütunun Adının Olduğu Tek Elemanlı Bir Liste Oluşur.
    actual_col = [col for col in data_frame.columns if col.lower() == x_col.lower()][0]

    # Ekran Görüntüsündeki Genel Ortalamayı (Örn: %20,4),Gelen Verinin Ağırlıklı Ortalamasından Hesaplar!
    # Grup Bazındaki Müşteri Sayılarını Topla → Toplam Müşteri Sayısını Bul->
    # → Churn Oranlarını Müşteri Sayılarıyla Ağırlıklandır → Genel Churn Oranını Bul.
    overall_avg = (data_frame["churn_rate"] * data_frame["customer_count"]).sum() / data_frame["customer_count"].sum()

    fig = px.bar(data_frame,
                x=actual_col,
                y="churn_rate",
                text="customer_count",
                title=title,
                labels={actual_col: x_col.upper(), "churn_rate": "Churn Rate (%)"})

    # Etiketleri Çubukların Üstüne Yerleştirir!
    fig.update_traces(textposition="outside")

    # Kesikli Genel Ortalama Çizgisi ve Etiketi!
    fig.add_hline(y=overall_avg,
                  line_dash="dash",
                  line_color="#334155",
                  annotation_text=f"Overall Average {overall_avg:.1f}%",
                  annotation_position="bottom right")
    return fig
```

```python
df_bank_churn["age_group"] = pd.cut(df_bank_churn["age"],
                                    bins=[17, 30, 40, 50, 60, 100],
                                    labels=["18-30", "31-40", "41-50", "51-60", "60+"])
churn_by_age = churn_by("age_group")
print(churn_by_age)
# 'Pd.Cut()' Fonksiyonu, Sürekli Sayısal Verileri Belirlenen Aralıklar Doğrultusunda Kategorik Gruplara Böler.
# 'Bins' Parametresi Aralık Sınırlarını Çizer, 'Labels' Parametresi İse Bu Gruplara Verilecek İsimleri Belirler.
# DF Bank Churn Data Frame'inde Age Group Sütunu Oluşturur.
```

```python
draw_bar_chart(churn_by_age, "age_group", "Churn Rate By Age Group (Label = Customer Count)").show()
```

```python
# Bakiyeyi Eşit Dilimlere Bölmek, Burada İşe Yaramaz: Müşterilerin %36'sı
# Tam Sıfırda Olduğu İçin Alt Dilimler Anlamsız Bölünür. Sabit Sınır Kullanılmalı.
df_bank_churn["balance_group"] = pd.cut(df_bank_churn["balance"],
                                        bins=[-1, 0, 100000, 150000, float("inf")],
                                        labels=["0_zero_balance", "1_medium_balance", "2_high_balance", "3_very_high_balance"])
#Float INF: Pozitif Sonsuzluğu İfade Eder. -1 ─────── 0 ───────────── 100K ───── 150K ───────── ∞
churn_by_balance = churn_by("balance_group")
print(churn_by_balance)
draw_bar_chart(churn_by_balance, "balance_group", "Churn Rate By Balance Group").show()
```

J. KORELASYON ANALİZİ: DEĞİŞKENLER ARASINDAKİ İLİŞKİ YOĞUNLUĞU!

```python
# Pearson Korelasyonu, İki Değişken Arasındaki Doğrusal İlişkinin Yönünü Ve Gücünü Ölçer.
# Değeri -1 İle +1 Arasındadır.
# -1 Mükemmel Negatif, 0 İlişki Yok, +1 Mükemmel Pozitif Doğrusal İlişkiyi İfade Eder.
# Sayısal Değişkenler Arasında Kullanılır.
# 0/1 Şeklinde Kodlanmış İkili Kategorik Değişkenler De Kullanılabilir.

# Spearman Korelasyonu, İki Değişken Arasındaki Monoton İlişkinin Yönünü Ve Gücünü Ölçer.
# Değeri -1 İle +1 Arasındadır.
# -1 Mükemmel Negatif, 0 İlişki Yok, +1 Mükemmel Pozitif Monoton İlişkiyi İfade Eder.
# Sayısal Ve Sıralı (Ordinal) Değişkenler Arasında Kullanılır.
# 0/1 Şeklinde Kodlanmış İkili Kategorik Değişkenler De Kullanılabilir.

# Mutual Information (MI), İki Değişken Arasındaki Genel Bağımlılığı Ölçer.
# Doğrusal Olmayan Ve Monoton Olmayan İlişkileri De Yakalayabilir.
# Değeri 0'dan Başlar Ve Üst Sınırı Yoktur.
# 0 Değeri, Değişkenler Arasında Bağımlılık Olmadığını İfade Eder.
# Sayısal-Sayısal, Sayısal-Kategorik, Kategorik-Sayısal Ve Kategorik-Kategorik Değişkenler Arasında Kullanılabilir.
# Değer Arttıkça Değişkenler Arasındaki Bağımlılık Gücü Artar.
```

```python
#Binary Kategorik → 0/1 ise Pearson, Spearman ve MI İle Analiz Edilebilir.
df_bank_churn["gender_encoded"] = df_bank_churn["gender"].map({"Male": 0,"Female": 1})#Same With Label Encoding!
```

```python
# Get Dummies(): Kategorik Değerleri One-Hot Encoding Yöntemiyle Ayrı 0/1 Sütunlarına Dönüştürür!
# Prefix, Sütunların Her Birinin Başına Country Ön Adını Ekler!
country_dummies = pd.get_dummies(df_bank_churn["country"],prefix="country",dtype=int)
df_bank_churn = df_bank_churn.join(country_dummies)
```

```python
# Pearson Korelasyon, Doğrusal İlişkileri Gösterir. Doğrusal Olmayan İlişkilerde Korelasyon Katsayısı
# Yanıltıcı Biçimde 0'a Yakın Çıkabilir Ve Var Olan Güçlü İlişkiyi Tamamen Gizleyebilir.
correlation = df_bank_churn.drop(columns=["customer_id"]).select_dtypes("number")
correlation_matrix = correlation.corr()["churn"].drop("churn").round(3).sort_values()
print(correlation_matrix)
# 'Churn' Sütununun Diğer Sayısal Değişkenlerle Olan Korelasyon Katsayılarını Hesaplar.
# Churn'ün Kendisiyle Olan Korelasyonunu (1.0) Listeden Çıkarır.
# Sonuçları Virgülden Sonra 3 Basamağa Yuvarlar.
# En Negatif İlişkiden En Pozitif İlişkiye Doğru Küçükten Büyüğe Sıralar.

px.imshow(correlation.corr().round(2), text_auto=True, aspect="auto",
          color_continuous_scale="RdBu_r", zmin=-1, zmax=1,
          title="Korelasyon Matrisi").show()
# Grafiği Oluşturan Kodda Tüm Değişkenlerin Arasındaki İlişkiye Bakılmakta.
# Sadece Churn İle İlişkilerine Değil.
```

```python
from sklearn.feature_selection import mutual_info_classif
#Sadece Pearson → Yaş İle Churn Arasında Düz Çizgi İlişkisi Var Mı?
#Spearman → Yaş Arttıkça Churn Genel Olarak Aynı Yönde Mi Hareket Ediyor?
#MI → Yaşı Bildiğimde, Müşterinin Churn Edip Etmeyeceği Hakkında Ne Kadar Bilgi Kazanıyorum?
#Örnek: Grafik: Yükseliş - Düşüş - Düşüş Yükseliş!
# Mutual Information, Değişkenler Arasındaki Doğrusal Olmayan İlişkiler Dahil Genel Bağımlılığı Ölçer!
X = df_bank_churn.drop(columns=["customer_id", "churn"]).select_dtypes("number")
y = df_bank_churn["churn"]
mi_scores = mutual_info_classif(X,y,random_state=42)
mi_scores = (pd.Series(mi_scores, index=X.columns).round(3).sort_values())
print(mi_scores)
# 'Churn' İle Diğer Sayısal Değişkenler Arasındaki Mutual Information Değerlerini Hesaplar.
# Sonuçları Virgülden Sonra 3 Basamağa Yuvarlar.
# Değişkenleri Mutual Information Değerine Göre Küçükten Büyüğe Sıralar.
```

```python
#Pearson = “Düz Bir Çizgi Gibi Mi İlerliyor?”
#Spearman = “Genel Olarak Aynı Yönde Mi İlerliyor?”
#Spearman Korelasyonu, Değişkenler Arasındaki Monoton İlişkileri Ölçer!
#Pozitif Monoton vs Negatif Monoton!
spearman = (df_bank_churn.drop(columns=["customer_id"])
            .select_dtypes("number")
            .corr(method="spearman")["churn"]
            .drop("churn")
            .round(3)
            .sort_values())
print(spearman)

#'Churn' İle Diğer Sayısal Değişkenler Arasındaki Spearman Korelasyonlarını Hesaplar.
#Sonuçları Virgülden Sonra 3 Basamağa Yuvarlar.
#En Negatif İlişkiden En Pozitif İlişkiye Doğru Sıralar.
```

```python
# Chi-Square Testi, İki Kategorik Değişken Arasında İstatistiksel Olarak Anlamlı Bir İlişki Olup Olmadığını Test Eder!
# Kategorik Feature İle Kategorik Target Arasındaki Bağımlılığı Değerlendirmek İçin Kullanılır!
# p-value < 0.05 İse Değişkenler Arasında İstatistiksel Olarak Anlamlı Bir İlişki Vardır!
from scipy.stats import chi2_contingency

categorical_columns = ["country", "gender", "credit_card", "active_member"]
for column in categorical_columns:
  contingency_table = pd.crosstab(df_bank_churn[column],df_bank_churn["churn"])
  p_value = chi2_contingency(contingency_table).pvalue
  print(f"{column}: p-value = {p_value:.4f}")
```

```python
# Cramér's V, İki Kategorik Değişken Arasındaki İlişkinin Gücünü Ölçer!
# Chi-Square Testi Anlamlı Bir İlişki Bulduğunda, Bu İlişkinin Ne Kadar Güçlü Olduğunu Değerlendirmek İçin Kullanılır!
# Değeri 0 İle 1 Arasındadır; 0 İlişki Olmadığını, 1 İse Mükemmel İlişkiyi İfade Eder!
from scipy.stats import chi2_contingency
import numpy as np

categorical_columns = ["country", "gender", "credit_card", "active_member"]
for column in categorical_columns:
    contingency_table = pd.crosstab(df_bank_churn[column],df_bank_churn["churn"])

    chi2_result = chi2_contingency(contingency_table)
    chi2 = chi2_result.statistic

    n = contingency_table.values.sum()
    min_dimension = min(contingency_table.shape) - 1
    cramers_v = np.sqrt(chi2 / (n * min_dimension))

    print(f"{column}: Cramér's V = {cramers_v:.3f}")
```
